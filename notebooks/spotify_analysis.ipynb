# Spotify Data Analysis

Exploratory analysis of a Spotify tracks dataset using Python, Pandas, SQL, and Plotly.

The analysis focuses on artist representation, popularity trends, explicit content across decades, and the relationship between song duration and popularity.

## Project Objective

The goal of this project is to analyze Spotify track data and identify patterns in music popularity, artist representation, explicit content, and song characteristics across different decades.

## Key Questions

- Which artists appear most frequently in the dataset?
- How has music representation changed across decades?
- How has explicit content changed over time?
- Are explicit tracks more popular than non-explicit tracks?
- Is track duration related to popularity?

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../data/spotify_analysis.csv")

df.head(10)
```

## Data Overview

The dataset used in this notebook contains the analysis-ready variables prepared during the data cleaning stage:

- track name
- artist name(s)
- album release year
- track duration
- explicit content flag
- popularity score

## Visual Analysis

### 1. Top 10 Artists by Number of Tracks

```python
top_artists = df["artist_names"].value_counts().head(10).reset_index()

top_artists
```

```python
fig = px.bar(
    top_artists,
    x="artist_names",
    y="count",
    title="Top 10 Artists by Number of Tracks",
    labels={
        "artist_names": "Artist",
        "count": "Number of Tracks"
    }
)

fig.show()
```

**Finding:** Taylor Swift has the highest number of tracks (50), followed by P!nk (47) and Elvis Presley (46), while the top 10 include both modern and classic artists.

### 2. Song Duration vs. Popularity

```python
df["duration_minutes"] = df["track_duration_ms"] / 60000

fig = px.scatter(
    df,
    x="duration_minutes",
    y="popularity",
    title="Song Duration vs. Popularity"
)

fig.show()
```

```python
correlation = df["duration_minutes"].corr(df["popularity"])

print(f"Correlation between song duration and popularity: {correlation:.3f}")
```

**Finding:** Most tracks are between 2 and 5 minutes long, with a few outliers. The correlation between duration and popularity is only 0.029, indicating almost no linear relationship between the two.

### 3. Explicit Tracks by Decade

```python
df["decade"] = (df["album_release_year"] // 10) * 10

df[["album_release_year", "decade"]].head(10)
```

```python
explicit_by_decade = (
    (df.groupby("decade")["explicit"].mean() * 100)
    .reset_index(name="explicit_percentage")
)

explicit_by_decade
```

```python
fig = px.line(
    explicit_by_decade,
    x="decade",
    y="explicit_percentage",
    title="Explicit Tracks by Decade",
    labels={
        "decade": "Decade",
        "explicit_percentage": "Explicit Tracks (%)"
    }
)

fig.show()
```

**Finding:** The share of explicit tracks has been increasing since the 1980s, with the strongest growth occurring after 2010.

### 4. Average Popularity by Decade

```python
popularity_by_decade = (
    df.groupby("decade")["popularity"]
    .mean()
    .reset_index()
)

popularity_by_decade["popularity"] = popularity_by_decade["popularity"].round(2)

popularity_by_decade
```

```python
fig = px.line(
    popularity_by_decade,
    x="decade",
    y="popularity",
    title="Average Popularity by Decade",
    labels={
        "decade": "Decade",
        "popularity": "Average Popularity"
    }
)

fig.show()
```

**Finding:** Average popularity generally declined from the 1970s to the 2000s, followed by a strong increase in the 2020s.

### Average Popularity — Explicit vs. Non-Explicit

```python
explicit_popularity = (
    df.groupby("explicit")["popularity"]
    .mean()
    .reset_index()
)

explicit_popularity["explicit"] = explicit_popularity["explicit"].map({
    True: "Explicit",
    False: "Non-Explicit"
})

fig = px.bar(
    explicit_popularity,
    x="explicit",
    y="popularity",
    title="Average Popularity: Explicit vs. Non-Explicit",
    labels={"explicit": "Track Type", "popularity": "Average Popularity"}
)

fig.show()
```

**Finding:** Explicit tracks have a higher average popularity score (39.85) than non-explicit tracks (32.09) in this dataset.

## Key Findings

- Taylor Swift has the highest number of tracks (50), followed by P!nk (47) and Elvis Presley (46).
- The share of explicit tracks has increased since the 1980s, reaching 13.95% in the 2020s.
- Average popularity generally declined from the 1970s to the 2000s, followed by a strong increase in the 2020s.
- Track duration has almost no linear relationship with popularity (r = 0.029).
- Explicit tracks have a higher average popularity score (39.85) than non-explicit tracks (32.09).
